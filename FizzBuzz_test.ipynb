# FIZZBUZZ TEST

```python
def fizzbuzz(n = 10):

  for number in range(1, n):
    if (number % 2 == 0) and (number % 3 == 0):
      print(str(number) + ': Fizzbuzz')
    elif (number % 2 == 0):
      print(str(number) + ': Fizz')
    elif (number % 3 == 0):
      print(str(number) + ': Buzz')
    else:
      print(str(number) + ': <>')
```

```python
def is_divisable(x, k):
  return x % k == 0

def get_answer(number, answer):
  return str(number) + ': ' + answer

def fizzbuzz(n = 10):

  for number in range(1, n):
    if is_divisible(number, 2) and is_divisible(number, 3):
      answer = get_answer(number, 'fizzbuzz')

    elif is_divisable(number, 2):
      answer = get_answer(number, 'fizz')

    elif is_divisable(number, 3):
      answer = get_answer(number, 'buzz')

    else:
      answer = get_answer(number, '<>')

    print(answer)
```

```python
answer = fizzbuzz(100)
```

functions created: is_divisible and get_answer.
