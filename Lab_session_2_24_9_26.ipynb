## Iteration
Iteration in Python is the process of accessing elements of an iterable (like lists, tuples, strings, sets, or dictionaries) one at a time, usually using loops such as for or while.

## Adding a loop

```python
primes = [2, 101, 5, 7, 11, 13, 17, 19]
s = 0
for prime in primes:
  print(prime)
```

This will: for each element in primes go through each element one by one and call those elements, get the first element then the second and so on. it will refer to element 1 then 2 then 3 and so on. This is called a for loop.

When you enter at the end of a line of code, it goes a tab in which means it is in the indentation, the iteration. This is asking fo each elements of prime, print each one out one by one, that's what a loop does, goes through each element one at a time.

```python
primes = [2, 101, 5, 7, 11, 13, 17, 19]
s = 0
for prime in primes:
  s = s + prime
  print (s)
```

S starts as 0 then + prime and the first prime 2 so the answer is 2
s is now 2 so its s + the next prime in the order which is 101 so it is 2 +101 so the answer is 103. now s is worth 103 and the next prime in the order is 5 so now it is s + 5 which is 103 + 5 which makes the answer 108.
