# conditionals

### are if something is true, then do this. mostly called IF statements

```python
numbers = [4, 10101, 8, 10, 3, 16, 17, 23, 567]
for number in numbers:
  if number <= 10:
    print (str(number) + ' is a Small number')
  else:
    print(str(number) + ' is not a small number')
```
